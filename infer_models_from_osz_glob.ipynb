# Infer Models From `.osz` Glob

Notebook layout:
1. Global imports + inputs
2. Context-model inference cell
3. Baseline-model inference cell
4. Baseline-maxopt inference cell

Defaults:
- `.osz` glob uses beachballs sample: `sample_data/raw/*2034220*.osz`
- checkpoints include context, baseline, and baseline maxopt

```python
from pathlib import Path

from src.inference.infer_from_osz import main as infer_from_osz_main

repo_root = Path.cwd()

# Default test input: beachballs set.
osz_glob = "sample_data/raw/*2034220*.osz"

work_root = repo_root / "sample_data" / "inference_work"
output_root = repo_root / "sample_data" / "inference_outputs" / "combined"

# Rhythm-preserving sampling defaults.
max_decode_len = 64
temperature = 1.0
top_p = 0.9
top_k = 8
ts_top_k = 16
min_event_candidates = 2
repetition_penalty = 1.0

context_ckpt = repo_root / "checkpoints" / "sample_large_context" / "last.ckpt"
baseline_ckpt = repo_root / "checkpoints" / "sample_large_baseline" / "last.ckpt"
baseline_maxopt_ckpt = repo_root / "checkpoints" / "sample_large_baseline_maxopt" / "last.ckpt"

print(f"osz_glob={osz_glob}")
print(f"work_root={work_root}")
print(f"output_root={output_root}")
print(f"context_ckpt={context_ckpt} | exists={context_ckpt.exists()}")
print(f"baseline_ckpt={baseline_ckpt} | exists={baseline_ckpt.exists()}")
print(f"baseline_maxopt_ckpt={baseline_maxopt_ckpt} | exists={baseline_maxopt_ckpt.exists()}")

def run_inference(checkpoint_path: Path):
    argv = [
        "--checkpoints", str(checkpoint_path),
        "--osz-inputs", osz_glob,
        "--work-root", str(work_root),
        "--output-root", str(output_root),
        "--max-decode-len", str(max_decode_len),
        "--temperature", str(temperature),
        "--top-p", str(top_p),
        "--top-k", str(top_k),
        "--ts-top-k", str(ts_top_k),
        "--min-event-candidates", str(min_event_candidates),
        "--repetition-penalty", str(repetition_penalty),
    ]
    print("Running inference with args:")
    print(" ".join(argv))
    rc = infer_from_osz_main(argv)
    print(f"done, rc={rc}")
```

```python
# Context model
run_inference(context_ckpt)
```

```python
# Baseline model
run_inference(baseline_ckpt)
```

```python
# Baseline maxopt model
run_inference(baseline_maxopt_ckpt)
```
